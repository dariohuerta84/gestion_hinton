```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Crear una imagen de prueba: un cuadrado azul con un círculo rojo
imagen = np.zeros((400, 400, 3), dtype="uint8")
imagen[:] = (255, 0, 0) # Fondo Azul (en BGR)
cv2.circle(imagen, (200, 200), 100, (0, 0, 255), -1) # Círculo Rojo central

# Convertir de BGR a RGB para que Matplotlib lo muestre bien
imagen_rgb = cv2.cvtColor(imagen, cv2.COLOR_BGR2RGB)

plt.imshow(imagen_rgb)
plt.title(f"OpenCV Versión: {cv2.__version__}")
plt.axis('off')
plt.show()
```
