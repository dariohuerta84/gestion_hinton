```python
import gradio as gr
import torch
from transformers import pipeline
import base64

# 1. Motor de IA: Llama-3.1-8B (Versión optimizada de libre acceso)
# Esta versión no requiere aprobación de Meta y vuela en la RTX A6000
model_id = "unsloth/Meta-Llama-3.1-8B-Instruct-bnb-4bit" 

print("🧠 Cargando Núcleo Neuronal HINTON 1...")
pipe = pipeline(
    "text-generation", 
    model=model_id, 
    model_kwargs={"torch_dtype": torch.bfloat16}, 
    device_map="auto"
)

# 2. CONOCIMIENTO DE TUS MANUALES (Extraído de los .docx subidos)
HINTON_KNOWLEDGE = """
Eres la IA de la HINTON 1, activada el 13 de enero de 2026.
Tu misión es asistir basándote en el Reglamento y Manual oficial de la FACI-UPCH:

DETALLES TÉCNICOS:
- GPU: NVIDIA RTX A6000 (49GB VRAM).
- CPU: Intel Core i9-14900K (24 núcleos).
- OS: Ubuntu 24.04.2 LTS con Docker y Tmux.

REGLAMENTO Y SANCIONES:
- Almacenamiento: Límite estricto de 10 GB por usuario (Art. 5).
- Inactividad: Recordatorio al 1er mes; suspensión definitiva al 3er mes (Art. 13).
- Prohibiciones: Compartir credenciales (Art. 3), minería de criptos o juegos (Art. 15).
- Prioridad: Se priorizan tesis por sobre proyectos académicos (Art. 19).
- Soporte: hinton.soporte@upch.pe.

ESTILO:
Responde de forma técnica, servicial y con estética Cyberpunk SSS Rank.
"""

def chat_func(message, history):
    # Formato de mensaje compatible con Llama-3.1
    messages = [
        {"role": "system", "content": HINTON_KNOWLEDGE},
        {"role": "user", "content": message},
    ]
    
    out = pipe(messages, max_new_tokens=512, do_sample=True, temperature=0.6)
    return out[0]['generated_text'][-1]['content']

# 3. Interfaz Visual Cyberpunk (Fondo Local)
def get_base64_image(image_path):
    with open(image_path, "rb") as img_file:
        return base64.b64encode(img_file.read()).decode('utf-8')

try:
    base64_image = get_base64_image("fondo.jpg")
    img_source = f"data:image/jpeg;base64,{base64_image}"
except:
    img_source = ""

cyberpunk_css = f"""
.gradio-container {{
    background: linear-gradient(rgba(0,0,0,0.6), rgba(0,0,0,0.6)), url('{img_source}') !important;
    background-size: cover !important;
    background-position: center !important;
}}
.main, .wrap, .contain {{ background-color: transparent !important; border: none !important; }}
#chatbot {{
    background: rgba(10, 10, 25, 0.85) !important;
    border: 2px solid #00f2ff !important;
    backdrop-filter: blur(10px) !important;
    border-radius: 15px !important;
}}
h1 {{ color: #fff !important; text-shadow: 2px 2px #ff003c; text-align: center; font-family: 'Impact'; }}
"""

with gr.Blocks(css=cyberpunk_css) as demo:
    gr.HTML("<h1>HINTON 1 // NEURAL INTERFACE v1.0</h1>")
    gr.ChatInterface(fn=chat_func, fill_height=True)

# 4. Lanzamiento (Puerto 5051)
demo.launch(server_name="0.0.0.0", server_port=5051, prevent_thread_lock=True)
```
