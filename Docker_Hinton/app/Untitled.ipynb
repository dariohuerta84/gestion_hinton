```python
import gradio as gr
import torch
from transformers import pipeline
import base64
import os

# 1. Configuración del Motor (Actualizado para evitar avisos de deprecación)
model_id = "unsloth/Meta-Llama-3.1-8B-Instruct-bnb-4bit" 

print("🧠 Cargando Núcleo Neuronal HINTON 1...")
pipe = pipeline(
    "text-generation", 
    model=model_id, 
    model_kwargs={"dtype": torch.bfloat16}, # Cambiado de torch_dtype a dtype
    device_map="auto"
)

# 2. Base de Conocimiento HINTON 1
HINTON_KNOWLEDGE = """
Eres la IA de la HINTON 1, activada el 13 de enero de 2026.
Tu misión es asistir basándote en el Reglamento y Manual oficial de la FACI-UPCH.
... (resto de tu prompt) ...
"""

def chat_func(message, history):
    messages = [
        {"role": "system", "content": HINTON_KNOWLEDGE},
        {"role": "user", "content": message},
    ]
    out = pipe(messages, max_new_tokens=512, do_sample=True, temperature=0.6)
    return out[0]['generated_text'][-1]['content']

# 3. Gestión de Imagen de Fondo (Rutas de Docker_Hinton)
def get_base64_image(image_path):
    if os.path.exists(image_path):
        with open(image_path, "rb") as img_file:
            return base64.b64encode(img_file.read()).decode('utf-8')
    return None

# Intentar rutas comunes en tu estructura de archivos
# Tu estructura muestra fondo.jpg en la raíz y en /app
img_data = get_base64_image("app/fondo.jpg") or get_base64_image("fondo.jpg")

if img_data:
    img_source = f"data:image/jpeg;base64,{img_data}"
    print("✅ Imagen de fondo cargada con éxito.")
else:
    img_source = ""
    print("⚠️ No se encontró fondo.jpg en /app/ o raíz.")

cyberpunk_css = f"""
.gradio-container {{
    background: linear-gradient(rgba(0,0,0,0.6), rgba(0,0,0,0.6)), url('{img_source}') !important;
    background-size: cover !important;
    background-position: center !important;
}}
#chatbot {{
    background: rgba(10, 10, 25, 0.85) !important;
    border: 2px solid #00f2ff !important;
    backdrop-filter: blur(10px) !important;
    border-radius: 15px !important;
}}
h1 {{ color: #fff !important; text-shadow: 2px 2px #ff003c; text-align: center; }}
"""

# 4. Interfaz (Ajustada para Gradio 6.0+)
with gr.Blocks() as demo:
    gr.HTML("<h1>HINTON 1 // NEURAL INTERFACE v1.0</h1>")
    gr.ChatInterface(fn=chat_func, fill_height=True)

# 5. Lanzamiento (Puerto 5051 con el CSS aquí para evitar UserWarning)
# Se usa server_name 0.0.0.0 para permitir acceso en la red UPCH [cite: 138, 296]
demo.launch(
    server_name="0.0.0.0", 
    server_port=5051, 
    prevent_thread_lock=True,
    css=cyberpunk_css  # Gradio 6.0 pide el CSS aquí
)
```
