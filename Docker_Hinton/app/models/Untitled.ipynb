```python
import torch
import os

# Definimos la ruta en el volumen montado para persistencia
MODEL_PATH = "/app/models/mi_modelo_sss.pth"

def save_checkpoint(model, optimizer, epoch, loss):
    """Guarda el estado completo del entrenamiento en el disco físico."""
    torch.save({
        'epoch': epoch,
        'model_state_dict': model.state_dict(),
        'optimizer_state_dict': optimizer.state_dict(),
        'loss': loss,
    }, MODEL_PATH)
    print(f"🔥 Checkpoint SSS Rank guardado en: {MODEL_PATH}")

# Para recuperar el progreso si reinicias el Docker
def load_checkpoint(model, optimizer):
    if os.path.exists(MODEL_PATH):
        checkpoint = torch.load(MODEL_PATH)
        model.load_state_dict(checkpoint['model_state_dict'])
        optimizer.load_state_dict(checkpoint['optimizer_state_dict'])
        print("⚡ Dominio Expandido restaurado: Modelo cargado con éxito.")
        return checkpoint['epoch'], checkpoint['loss']
    return 0, None
```
